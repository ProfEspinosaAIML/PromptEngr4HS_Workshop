**Workshop Theme:**  
Using AI *without* blindly trusting it

Today is not about getting the “right answers.”

### Why are we doing this?

Most people learn how to:
- Ask AI homework questions  
- Generate summaries  
- Get quick explanations  

Very few people learn how to:
- Push AI until it breaks  
- Spot when it *sounds* right but isn’t  
- Ask follow-up questions that reveal gaps, bias, or hallucinations  

That’s what this workshop is about.


### By the end of this workshop, you should be able to:

✅ Write prompts that guide AI thinking (not just answers)  
✅ Explore multiple perspectives on the same topic  
✅ Recognize when AI may be hallucinating  
✅ Use images and visuals generated by AI *critically*, not passively  
✅ Decide when AI is helpful — and when it shouldn't be trusted

### Ground Rules (Very Important)

1. **AI sounds confident — even when it’s wrong**
2. **A polished answer is not the same as a true answer**
3. **You are allowed to question AI**
4. **You are allowed to be wrong — that’s how this works**
5. **Hallucinations are not “bugs” — they are learning opportunities**

If something feels off, weird, or too perfect…  
That’s exactly where we want to be.


### How the next 50 minutes will work

We will follow this pattern:

**Teach → Create → Debate**

- **Teach:** See AI succeed — then fail — in real time  
- **Create:** Use prompts and images to explore ideas  
- **Debate:** Decide what to trust, what to question, and why  

You don’t need:
- Coding skills  
- Technical AI knowledge  
- Perfect prompts  

You *do* need:
- Curiosity  
- Skepticism  
- Willingness to ask follow-up questions


### Quick Warm‑Up (Answer Honestly)

Take 1–2 minutes to answer:

**1. What do you usually trust AI for?**  
(e.g., explanations, summaries, brainstorming, homework help)

**2. What is one thing you already don’t fully trust AI to do?**

There are no right or wrong answers here.


### One Last Thing Before We Start

At some point today, the AI will:
- Make things up  
- Invent experts  
- Give answers that *sound* smart but aren’t verifiable  

When that happens, don’t panic.

That moment is the **real lesson**.
