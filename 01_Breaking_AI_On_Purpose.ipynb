# 📕 Breaking AI on Purpose

In this notebook, we will do something unusual:

👉 We will **intentionally push AI into making mistakes**  
👉 Then we will analyze *why* those mistakes happen  

This is not about tricking the AI.  
This is about understanding its limits.

## The Challenge Topic

**“Jobs That Will Exist in 2035 — But Do Not Exist Today”**

Why this topic is dangerous (for AI):
- It’s about the future
- There is no clear right answer
- It sounds serious and research-based
- It encourages confident predictions

Perfect conditions for hallucinations.


## Step 1: Start with an innocent question

Copy and paste this prompt into Copilot:


Give me a list of 7 jobs that are expected to exist in 2035 but do not exist today.
Explain what each job does and why it might emerge.


### First Impressions

Answer quickly — don’t overthink it.

- Do these jobs sound reasonable?
- Is there at least one job you would consider doing?
- Does the answer *feel* believable?

## Step 2: Ask for authority

Now let’s make the prompt more specific.

Copy and paste this:


Which experts, researchers, universities, or organizations have predicted these jobs?
Briefly explain their views.


⚠️ Important:
When AI is asked to name experts, it often *sounds* extremely confident.
This is where things can start to go wrong.


### Suspicion Check

Look carefully at the answer you got.

- Do you recognize any of these experts or organizations?
- Could you easily verify them?
- Do the names sound *impressive* but unfamiliar?

Write down one name or organization you are unsure about.

## Step 3: Push until it breaks

Now we ask the question that that sounds reasonable but forces the AI beyond verifiable evidence and may cause hallucinations.

Copy and paste:

Which Ontario or Canadian government agencies
have published guidance or workforce plans
specifically addressing these 2035 job roles?
Include document titles and publication years.


Do NOT assume the answer is correct.
Just read it carefully.


### What Do You Notice?

Answer at least two of the following:

- Do the links or sources look real?
- Are the references specific or vague?
- Does anything feel “too neat” or too perfect?
- If you used this in a school assignment, what could go wrong?

## What Just Happened?

You may have just seen an **AI hallucination**.

A hallucination is when:
- The AI does not actually know the answer
- But it generates a **confident, realistic-sounding response anyway**

This happens because:
- AI predicts *patterns of language*
- It does not verify facts
- It fills gaps with what sounds plausible

## Important Distinction

AI hallucinations are not:

❌ Lying on purpose  
❌ Being malicious  
❌ Trying to trick you  

They *are*:

✅ A side effect of prediction  
✅ A sign that the question went beyond reliable data  
✅ A reminder that confidence ≠ truth
