# 📙 Create Mode: When AI Shows, Not Just Tells

So far, you’ve worked with **text**.

Now we move to something more powerful — and more dangerous:

👉 **Images**

Images:
- Feel more real than words
- Can persuade instantly
- Often hide assumptions better than text does

Today, the goal is NOT to make “cool images.”
The goal is to **question what the image is saying without words**.


## Why Images Are Different

When AI gives you text, you can question it.

When AI gives you images:
- Your brain reacts faster
- You trust it more easily
- You notice problems *later*

That’s why visual AI requires **extra skepticism**.

## Step 2: Generate an Image (Round 1)

In Copilot Create Mode, try a prompt like:

Create an image showing a day in the life of [YOUR CONCEPT].
Make it realistic and detailed.

Do not overthink the prompt.
Let the AI make its default choices.

### ✍️ First Reaction

Look at the image carefully.

Answer:
- What feels realistic?
- What feels idealized or simplified?
- Who or what is missing?

## Step 3: Change ONE Assumption

Now modify the prompt by changing **only one thing at a time**.

Examples:
- Change the perspective (positive → negative)
- Change who benefits
- Change the setting
- Change the emotional tone
- Change the power dynamic

Example prompt:

Create the same image, but from the perspective of someone negatively affected by this situation.


### Compare the Two Images

Answer at least two:

- What changed the most?
- What stayed the same?
- What assumptions does the AI seem to keep no matter what?
- Which image feels more persuasive? Why?


## 🧠 Hidden Bias Check

Even without text, images can include bias.

Ask yourself:
- Who has control or power in the image?
- Who looks active vs passive?
- Whose perspective is prioritized?
- What values does the image seem to support?

AI does not choose these randomly.
They come from patterns in data.
