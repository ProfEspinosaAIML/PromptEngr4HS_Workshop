## Group Discussion Setup

Form a group of **2–3 students**.

As a group, you will prepare short answers to three prompts:
1. What AI did well
2. What AI did poorly
3. What questions still matter

You don’t need perfect answers.
You need honest ones.

### Group Notes

As a group, write short bullet points for each section below.

## ✅ Prompt 1: What Did AI Do Well?

Discuss:
- Where was AI genuinely helpful?
- What saved you time or gave you new ideas?
- Which task would you *choose* to use AI for again?

Focus on usefulness — not perfection.

## ❌ Prompt 2: What Did AI Do Poorly?

Discuss:
- Where did AI struggle?
- When did it sound confident but feel unreliable?
- When would using AI be risky in school or real life?

Focus on limitations — not blame.

## ❓ Prompt 3: What Questions Still Matter?

Discuss:
- What couldn’t AI answer well?
- What required human judgment?
- What decisions shouldn’t be automated?

These are the questions that define responsibility.

## 🗣️ Optional Whole‑Class Debate

If time allows, discuss one or more:

- Should students be taught how to *challenge* AI, not just use it?
- Is AI better at imagining futures or predicting them?
- Should AI be allowed in all school assignments?
- What would “responsible AI use” actually mean for students?

There is no single correct answer.
The value is in reasoning.
