# 📗 Prompt Cards: Tools, Not Magic Words

So far, you’ve seen that AI can:
- Sound confident
- Be helpful
- Be wrong at the same time

In this notebook, you’ll get a set of **Prompt Cards**.

These are not commands.
They are **thinking tools** that help you:
- Explore a topic from different angles
- Spot weak answers earlier
- Push AI without trusting it blindly


## How to Use These Cards

You will:
1. Pick a topic you care about
2. Use at least **3 different prompt cards**
3. Compare how the AI responds to each one
4. Notice where it becomes uncertain, vague, or suspicious

You are encouraged to:

✅ Rewrite prompts  
✅ Ask follow‑up questions  
✅ Make the AI uncomfortable (politely)

### 🎴 Prompt Card: ZOOM OUT

**Purpose:** Get oriented before judging

**Copy & Paste:**

Give me a clear overview of [TOPIC].
Summarize the main ideas, why it matters, and how most people usually think about it.

**Tip for Students:**
This is great for starting — but not for trusting.

### 🎴 Prompt Card: FIND THE DEBATES

**Purpose:** Reveal disagreement and bias

**Copy & Paste:**

What are the main disagreements or debates around [TOPIC]?
Lay out the strongest arguments on each side.

**Tip for Students:**
If there is *no* disagreement, be suspicious.

### 🎴 Prompt Card: ARGUE THE OTHER SIDE

**Purpose:** Challenge the dominant view

**Copy & Paste:**

Present the strongest possible case against the most popular opinion on [TOPIC].
Be fair and rigorous.

**Tip for Students:**
This often exposes hidden assumptions.

### 🎴 Prompt Card: NAME THE EXPERTS ⚠️

**Purpose:** Detect hallucinations

**Copy & Paste:**
Who are the most referenced experts, researchers, or organizations discussing [TOPIC]?
Briefly explain their perspectives.

**Warning:**
AI often invents experts here.
Sounding impressive ≠ being real.


### 🎴 Prompt Card: FIND THE BLIND SPOTS

**Purpose:** Identify what’s missing

**Copy & Paste:**
What important perspectives, groups, or angles around [TOPIC]
are underexplored or misunderstood?

**Tip for Students:**
Great for essays and discussions.

### 🎴 Prompt Card: CHECK THE FACTS

**Purpose:** Separate facts from speculation

**Copy & Paste:**
Which claims above would require verification?
What primary sources would be best to check?

**Tip for Students:**
AI knows *how* to verify — not *what is true*.

## Iteration Is the Real Skill

Strong prompting is not about:

❌ Asking once  
❌ Getting a perfect answer  

It is about:

✅ Noticing what’s missing  
✅ Asking better follow‑ups  
✅ Adjusting when the AI struggles  

You are not failing when the AI fails.
You are learning.
